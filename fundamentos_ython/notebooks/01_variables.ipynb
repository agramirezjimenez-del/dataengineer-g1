### CASTEO DE VARIABLES

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))

precio = "130.5"
precio = float(precio)
print(type(precio))
print(precio)

precio =  int(precio)
print(precio)
```

```python
activo = 0
print(activo)
activo = bool(activo)
print(activo)
```
