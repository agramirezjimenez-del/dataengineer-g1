### Pregunta 1.3: Contar frecuencia de elementos

Dada una lista de nombres repetidos, muestra **cuántas veces aparece cada nombre**.
Cada nombre debe mostrarse una sola vez, en el orden en que aparece por primera vez.

```python
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'María', 'Carlos']
```

```python
#Frecuencia de cada nombre
frecuencia_nombres = {}
for nombre in nombres:
    frecuencia_nombres[nombre] = frecuencia_nombres.get(nombre, 0) + 1

for nombre, cantidad in frecuencia_nombres.items():
    print(f"{nombre}: {cantidad}")
```

```python
numeros = [1, 2, 3, 4]
letras = ['a', 'b', 'c', 'd']
```

### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí
inventario_total = {}

inventarios = [inventario_norte, inventario_sur, inventario_centro]

articulos = set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys(),
])

print(articulos)
print("="*50)

for articulo in articulos:
    inventario_total[articulo] = {
        "cantidad_total": sum([inventario.get(articulo, 0) for inventario in inventarios]),
        "sucursales": sum([1 for inventario in inventarios if articulo in inventario])
    }

for articulo in inventario_total:
    print(f"articulo: {articulo} | {inventario_total[articulo]}")
    print('='*50)
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
import json

agrupados_por_departamento = {}

for empleado in empleados:
    dpto = empleado["departamento"]
    nivel = empleado["nivel"]
    
    agrupados_por_departamento.setdefault(dpto, {}).setdefault(nivel,{"nombres": [], "salarios": []})
    agrupados_por_departamento[dpto][nivel]["nombres"].append(empleado["nombre"])
    agrupados_por_departamento[dpto][nivel]["salarios"].append(empleado["salario"])
    
# Reemplazar las listas de salarios por el promedio y agregar cantidad
for dpto, niveles in agrupados_por_departamento.items():
    for nivel, datos in niveles.items():
        datos["cantidad"] = len(datos["nombres"])
        datos["salario_promedio"] = sum(datos["salarios"]) / datos["cantidad"]
        del datos["salarios"]

print(json.dumps(agrupados_por_departamento, indent=2, ensure_ascii=False))
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
conteo = {c: 0 for c in candidatos}
invalidos = 0

for voto in votos:
    voto_limpio = voto.strip().capitalize() if voto else ""
    if voto_limpio in candidatos:
        conteo[voto_limpio] += 1
    else:
        invalidos += 1

total_validos = sum(conteo.values())

print("Conteo por candidato:")
for c, n in conteo.items():
    porcentaje = (n / total_validos * 100) if total_validos else 0
    print(f"  {c}: {n} votos ({porcentaje:.1f}%)")

print(f"Votos inválidos: {invalidos}")

# Ganador o empate
max_votos = max(conteo.values())
ganadores = [c for c, n in conteo.items() if n == max_votos]

if len(ganadores) == 1:
    print(f"Ganador: {ganadores[0]} con {max_votos} votos")
else:
    print(f"Empate entre: {', '.join(ganadores)} con {max_votos} votos cada uno")
```

### Pregunta 3.5: Índice invertido de documentos

Crea un **índice invertido** que permita saber en qué documentos aparece cada palabra.

**Requisitos:**
- Cada palabra debe apuntar a un **conjunto** (`set`) de identificadores de documentos.
- Ignora las palabras de una sola letra.
- Usa el índice para responder: ¿en qué documentos aparecen `"python"` y `"datos"`?

```python
documentos = {
    "doc1": "python y ciencia de datos",
    "doc2": "python para desarrollo web",
    "doc3": "ciencia de datos con inteligencia artificial",
}
```

```python
indice = {}

for doc_id, texto in documentos.items():
    palabras = texto.lower().split()
    for palabra in palabras:
        if len(palabra) <= 1:
            continue
        indice.setdefault(palabra, set()).add(doc_id)

# Consulta
for termino in ["python", "datos"]:
    print(f"'{termino}' aparece en: {indice.get(termino, set())}")
```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(ventas: list[float]) -> dict[str, float] | str:
    if not ventas:
        return "ERROR: No existen ventas"

    for venta in ventas:
        if venta < 0:
            return f"ERROR: Valor de venta negativo: {venta}"

    total_vendido = sum(ventas)
    promedio_vendido = total_vendido / len(ventas)
    venta_max = max(ventas)
    venta_min = min(ventas)

    ventas_superiores_promedio = sum([1 for venta in ventas if venta >= promedio_vendido])
    ventas_por_debajo_100 = sum([1 for venta in ventas if venta < 100])

    estadisticos = {
        "total_vendido": total_vendido,
        "promedio_vendido": promedio_vendido,
        "venta_max": venta_max,
        "venta_min": venta_min,
        "ventas_superiores": ventas_superiores_promedio,
        "ventas_por_debajo_100": ventas_por_debajo_100
    }

    return estadisticos


stats = analizar_ventas(ventas)
print(stats)
```

```python
stats = analizar_ventas([])
print(stats)
```

```python
stats = analizar_ventas([150.50, 200, 80.90, 300, 175.50, 90, -100])
print(stats)
```

### Pregunta 5.2: Validar una contraseña y explicar los errores

Crea una función `validar_password(password)` que determine si una contraseña cumple todas las reglas de seguridad.

**Requisitos:**
- Debe tener al menos 8 caracteres.
- Debe contener una letra mayúscula, una minúscula y un número.
- Debe incluir al menos uno de estos símbolos: `@`, `#`, `$` o `%`.
- No debe contener espacios.
- Debe retornar una tupla con el formato `(es_valida, errores)`, donde `errores` sea una lista.

```python
password_prueba = 'Python123'
```

```python
def validar_password(password: str) -> tuple[bool, list[str]]:
    errores = []

    if len(password) < 8:
        errores.append("Debe tener al menos 8 caracteres.")
    if not any(c.isupper() for c in password):
        errores.append("Debe contener al menos una letra mayúscula.")
    if not any(c.islower() for c in password):
        errores.append("Debe contener al menos una letra minúscula.")
    if not any(c.isdigit() for c in password):
        errores.append("Debe contener al menos un número.")
    if not any(c in "@#$%" for c in password):
        errores.append("Debe incluir al menos uno de estos símbolos: @, #, $, %.")
    if " " in password:
        errores.append("No debe contener espacios.")

    return (len(errores) == 0, errores)


password_prueba = 'Python123'
es_valida, errores = validar_password(password_prueba)
print(f"¿Válida? {es_valida}")
for e in errores:
    print(f" - {e}")
```

- Ordenar de mayor a menor promedio y, en caso de empate, alfabéticamente.

```python
empleados = {
    'Ana': [80, 90, 75],
    'Carlos': [60, 70, 65],
    'Lucía': [95, 85, 90],
    'Pedro': [70, 70, 70],
}
```

```python
def evaluar_empleados(empleados: dict, minimo: float = 70) -> list[tuple]:
    resultados = []

    for nombre, puntajes in empleados.items():
        promedio = sum(puntajes) / len(puntajes)

        if promedio >= 90:
            clasificacion = "Excelente"
        elif promedio >= 80:
            clasificacion = "Bueno"
        elif promedio >= 70:
            clasificacion = "Regular"
        else:
            clasificacion = "Deficiente"

        if promedio >= minimo:
            resultados.append((nombre, round(promedio, 2), clasificacion))

    # Ordenar por promedio desc, y alfabéticamente en empate
    resultados.sort(key=lambda x: (-x[1], x[0]))
    return resultados

for nombre, prom, clasif in evaluar_empleados(empleados):
    print(f"{nombre}: {prom} - {clasif}")
```

### Pregunta 5.4: Comprimir y descomprimir texto

Implementa dos funciones: `comprimir_texto(texto)` y `descomprimir_texto(texto)`.

**Requisitos:**
- `comprimir_texto` debe reemplazar caracteres consecutivos repetidos por el carácter y su cantidad.
  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaabbccccdaa'
```

```python
def comprimir_texto(texto: str) -> str:
    if not texto:
        return ""
    resultado = []
    actual = texto[0]
    contador = 1
    for c in texto[1:]:
        if c == actual:
            contador += 1
        else:
            resultado.append(f"{actual}{contador}")
            actual = c
            contador = 1
    resultado.append(f"{actual}{contador}")
    return "".join(resultado)


def descomprimir_texto(texto: str) -> str:
    if not texto:
        return ""
    resultado = []
    i = 0
    while i < len(texto):
        caracter = texto[i]
        if not caracter.isalpha():
            return "ERROR: formato comprimido inválido."
        i += 1
        num_str = ""
        while i < len(texto) and texto[i].isdigit():
            num_str += texto[i]
            i += 1
        if not num_str:
            return "ERROR: formato comprimido inválido."
        resultado.append(caracter * int(num_str))
    return "".join(resultado)


comprimido = comprimir_texto(texto)
print("Comprimido:", comprimido)
print("Descomprimido:", descomprimir_texto(comprimido))
print("Caso a12:", descomprimir_texto("a12"))
print("Inválido:", descomprimir_texto("abc"))
```

### Pregunta 5.5: Sistema de reservas de asientos

Crea una función `reservar_asientos(sala, asientos_solicitados)` que permita reservar varios asientos de una sala.

**Requisitos:**
- La sala será una lista de listas donde `L` significa libre y `O` significa ocupado.
- Cada asiento solicitado se expresará mediante una tupla `(fila, columna)`, **numeradas desde 1**.
- Validar que todas las posiciones existan y que los asientos estén libres.
- Realizar la reserva únicamente si todos están disponibles.
- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
def reservar_asientos(sala: list[list[str]], asientos_solicitados: list[tuple]) -> dict:
    # Validar que todos los asientos existan y estén libres
    for fila, col in asientos_solicitados:
        if not (1 <= fila <= len(sala)):
            return {"estado": f"ERROR: fila {fila} fuera de rango.",
                    "reservados": [], "sala": sala}
        if not (1 <= col <= len(sala[0])):
            return {"estado": f"ERROR: columna {col} fuera de rango.",
                    "reservados": [], "sala": sala}
        if sala[fila - 1][col - 1] == 'O':
            return {"estado": f"ERROR: asiento ({fila},{col}) ocupado.",
                    "reservados": [], "sala": sala}

    # Si todo está OK, reservar (sin modificar la original)
    nueva_sala = [fila[:] for fila in sala]
    for fila, col in asientos_solicitados:
        nueva_sala[fila - 1][col - 1] = 'O'

    return {"estado": "OK", "reservados": asientos_solicitados, "sala": nueva_sala}


def mostrar_sala(sala: list[list[str]]) -> None:
    print("    " + " ".join(f"C{j+1}" for j in range(len(sala[0]))))
    for i, fila in enumerate(sala, start=1):
        print(f"F{i}: " + "  ".join(fila))

resultado = reservar_asientos(sala, asientos)
print("Estado:", resultado["estado"])
print("Reservados:", resultado["reservados"])
print("Sala original intacta:")
mostrar_sala(sala)
print("Sala actualizada:")
mostrar_sala(resultado["sala"])
```

- Método `depositar(monto)`: suma al saldo, solo si el monto es positivo.
- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
cuenta1 = CuentaBancaria("Cesar Vallejo", 500)
cuenta2 = CuentaBancaria("Abraham Valdelomar", 1000)
cuenta3 = CuentaBancaria("Marie Curie", 100)

cuenta1.depositar(200)
cuenta1.retirar(100)

cuenta2.depositar(500)
cuenta2.retirar(300)

cuenta3.depositar(100)
cuenta3.retirar(50)

```

```python
class CuentaBancaria:
    def __init__(self, titular: str, saldo: float = 0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, monto: float) -> None:
        if monto > 0:
            self.saldo += monto
            print(f"Depósito exitoso. \n > Nuevo saldo: ${self.saldo:.2f}")
        else:
            print("ERROR: el monto debe ser positivo.")

    def retirar(self, monto: float) -> None:
        if monto <= 0:
            print("ERROR: el monto debe ser positivo.")
        elif monto > self.saldo:
            print("Fondos insuficientes.")
        else:
            self.saldo -= monto
            print(f"Retiro exitoso. \n > Nuevo saldo: ${self.saldo:.2f}")

    def consultar_saldo(self) -> float:
        return self.saldo




print(f"Titular: {cuenta1.titular}")

cuenta1.depositar(250)
cuenta1.retirar(120)
cuenta1.retirar(500)

print(f"-> Saldo final: ${cuenta1.consultar_saldo():.2f}")

print("="*50)

print(f"Titular: {cuenta2.titular}")

cuenta2.depositar(250)
cuenta2.retirar(120)
cuenta2.retirar(500)

print(f"-> Saldo final: ${cuenta2.consultar_saldo():.2f}")

print("="*50)
print(f"Titular: {cuenta3.titular}")

cuenta3.depositar(250)
cuenta3.retirar(120)
cuenta3.retirar(500)

print(f"-> Saldo final: ${cuenta3.consultar_saldo():.2f}")

```

```python
class Motor:
    def __init__(self, tipo: str, caballos: int):
        self.tipo = tipo
        self.caballos = caballos
        self.encendido = False

    def arrancar(self) -> str:
        if self.encendido:
            return f"El motor {self.tipo} ya estaba encendido."
        self.encendido = True
        return f"Motor {self.tipo} de {self.caballos} HP arrancado."


class Vehiculo:
    def __init__(self, marca: str, motor: Motor):
        self.marca = marca
        self.motor = motor

    def arrancar(self) -> str:
        return f"{self.marca}: {self.motor.arrancar()}"


```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python

bib = Biblioteca()
bib.agregar_libro(Libro("Cien años de soledad", "García Márquez"))
bib.agregar_libro(Libro("1984", "George Orwell"))
bib.agregar_libro(Libro("El principito", "Saint-Exupéry"))

ana = Usuario("Ana")
luis = Usuario("Luis")

bib.catalogo()
print()

bib.prestar("1984", ana)
bib.prestar("1984", luis)       
bib.prestar("El Quijote", ana)  
bib.devolver("1984", luis)      
bib.devolver("1984", ana)       

print()
bib.catalogo()
print()
print(f"Libros de Ana: {ana.libros_prestados}")
print(f"Libros de Luis: {luis.libros_prestados}")
```

```python
class Libro:
    def __init__(self, titulo: str, autor: str):
        self.titulo = titulo
        self.autor = autor
        self.disponible = True

    def __repr__(self):
        estado = "disponible" if self.disponible else "prestado"
        return f"Libro('{self.titulo}', '{self.autor}', {estado})"


class Usuario:
    def __init__(self, nombre: str):
        self.nombre = nombre
        self.libros_prestados = []

    def __repr__(self):
        return f"Usuario('{self.nombre}', prestados={len(self.libros_prestados)})"


class Biblioteca:
    def __init__(self):
        self.libros = []

    def agregar_libro(self, libro: Libro) -> None:
        self.libros.append(libro)

    def buscar(self, titulo: str) -> Libro | None:
        for libro in self.libros:
            if libro.titulo.lower() == titulo.lower():
                return libro
        return None

    def prestar(self, titulo: str, usuario: Usuario) -> None:
        libro = self.buscar(titulo)
        if libro is None:
            print(f"ERROR: el libro '{titulo}' no existe en la biblioteca.")
        elif not libro.disponible:
            print(f"ERROR: el libro '{titulo}' ya está prestado.")
        else:
            libro.disponible = False
            usuario.libros_prestados.append(libro)
            print(f"'{titulo}' prestado a {usuario.nombre}.")

    def devolver(self, titulo: str, usuario: Usuario) -> None:
        libro = self.buscar(titulo)
        if libro is None:
            print(f"ERROR: el libro '{titulo}' no existe en la biblioteca.")
        elif libro not in usuario.libros_prestados:
            print(f"ERROR: {usuario.nombre} no tiene el libro '{titulo}'.")
        else:
            libro.disponible = True
            usuario.libros_prestados.remove(libro)
            print(f"'{titulo}' devuelto por {usuario.nombre}.")

    def catalogo(self) -> None:
        print("Catálogo de la biblioteca:")
        for libro in self.libros:
            estado = "Disponible" if libro.disponible else "Prestado"
            print(f"  - {libro.titulo} ({libro.autor}) [{estado}]")

```
