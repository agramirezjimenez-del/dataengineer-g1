```python
import pandas as pd

df = pd.read_csv("../data/clientes_sucio.csv")
df.head(15)
```

```python
df.info()
```

### Encontrar duplicados

```python
filtro = df.duplicated()
df[filtro]
```

```python
clientes_sin_duplicados = df.drop_duplicates()
clientes_sin_duplicados.head(10)
```

```python
clientes_sin_duplicados["email"] = clientes_sin_duplicados["email"].str.strip().str.lower()
```

```python
clientes_sin_duplicados.head(10)
```

```python
clientes_sin_duplicados["nombre"] = clientes_sin_duplicados["nombre"].str.title().str.strip()
clientes_sin_duplicados["ciudad"] = clientes_sin_duplicados["ciudad"].str.title().str.strip()
clientes_sin_duplicados.head(10)
```

```python
### PREGUNTA DE LA CLASE PARA TRATAMIENTO DE CORREOS, AD-HOC A LA EMPRESA Y AL DOMINIO
'''
correo = "JAVIER.SANZ@mail.com" # asi es como te llega
lower_correo = "javier.sanz@mail.com" # Problema
esperado = "Javier.Sanz@mail.com"

splits = lower_correo.split("@")
nickname = splits[0].replace(".", " ").title().replace(" ", ".")
print(nickname + "@" + splits[1])
'''
```

```python
clientes_sin_duplicados["fecha_nac"] = pd.to_datetime(
    clientes_sin_duplicados["fecha_nac"],
    format="mixed"
)

clientes_sin_duplicados
```

```python
### DARLE TRATAMIENTO A UNA COLUMNA DE TEXTO PARA CONVERTIRLA A FORMATO NUMERICO Y PODER REALIZAR OPERACIONES
### USANDO MAP Y CASTEO A FLOAT64
def conversor_apunto_flotante(x: str) -> str:
    if x:
        return str(x).replace(".", "").replace(",", ".")
    return x

# clientes_sin_duplicados["gasto_total"].map(conversor_apunto_flotante).astype("float64")
```

```python
### USANDO UNA FUNCIÓN ANÓNIMA O FUNCIÓN LAMBDA
# clientes_sin_duplicados["gasto_total"].map(lambda x: str(x).replace(".", "").replace(",", "."))
```

```python
clientes_sin_duplicados["gasto_total"] = clientes_sin_duplicados["gasto_total"] \
    .str.replace(".", "") \
    .str.replace(",", ".").astype(pd.Float64Dtype())
```

```python
clientes_sin_duplicados
```

```python
clientes_sin_duplicados["activo"] = clientes_sin_duplicados["activo"].str.lower().str.replace(
    {"si": "1", "no": "0"}
).astype(int).astype(bool)

clientes_sin_duplicados
```

```python
clientes_sin_duplicados["telefono"] = clientes_sin_duplicados["telefono"].str.replace(r"\D", "", regex=True)
clientes_sin_duplicados
```

```python
clientes_sin_duplicados.to_csv("../data/clientes_limpios.csv", index=False)
```
