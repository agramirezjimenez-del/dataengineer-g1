### Ejercicio 15. Detalle general de pedidos

**Nivel:** Básico &nbsp;|&nbsp; **Conceptos:** JOIN de tres tablas

**Pregunta:** Lista cada pedido mostrando `order_id`, fecha, estado, canal, nombre del cliente y nombre del empleado responsable.

````python
%sql
SELECT o.order_id,
       o.order_date,
       o.status,
       o.channel,
       c.full_name AS cliente,
       e.full_name AS empleado
FROM pedidos o
INNER JOIN clientes  c ON o.customer_id = c.customer_id
INNER JOIN empleados e ON o.employee_id = e.employee_id
ORDER BY o.order_id;
````

> En estos ejercicios un pedido **no cancelado** es aquel cuyo `status` es distinto de `'cancelado'`.

### Ejercicio 16. Importe neto por pedido

**Nivel:** Intermedio &nbsp;|&nbsp; **Conceptos:** JOIN, SUM, fórmula de descuento, GROUP BY

**Pregunta:** Calcula el importe neto de cada pedido **no cancelado**. Cada línea vale `quantity * unit_price * (1 - discount_pct / 100)`. Muestra `order_id`, cliente, fecha y `net_order_amount`, redondeado a dos decimales.

````python
%sql
SELECT o.order_id,
       c.full_name AS cliente,
       o.order_date,
       ROUND(SUM(d.quantity * d.unit_price * (1 - d.discount_pct / 100)), 2) AS net_order_amount
FROM pedidos o
INNER JOIN clientes c         ON o.customer_id = c.customer_id
INNER JOIN detalle_pedidos d  ON o.order_id = d.order_id
WHERE o.status <> 'cancelado'
GROUP BY o.order_id, c.full_name, o.order_date
ORDER BY o.order_id;
````

### Ejercicio 17. Clientes frecuentes y su gasto

**Nivel:** Intermedio &nbsp;|&nbsp; **Conceptos:** JOIN, GROUP BY, HAVING, COUNT DISTINCT

**Pregunta:** Encuentra los clientes que tienen más de dos pedidos **no cancelados**. Muestra su nombre, número de pedidos y gasto neto total, ordenando desde el mayor gasto.

````python
%sql
SELECT c.customer_id,
       c.full_name AS cliente,
       COUNT(DISTINCT o.order_id) AS num_pedidos,
       ROUND(SUM(d.quantity * d.unit_price * (1 - d.discount_pct / 100)), 2) AS gasto_neto_total
FROM clientes c
INNER JOIN pedidos o         ON c.customer_id = o.customer_id
INNER JOIN detalle_pedidos d ON o.order_id = d.order_id
WHERE o.status <> 'cancelado'
GROUP BY c.customer_id, c.full_name
HAVING COUNT(DISTINCT o.order_id) > 2
ORDER BY gasto_neto_total DESC;
````

### Ejercicio 18. Productos nunca vendidos

**Nivel:** Intermedio &nbsp;|&nbsp; **Conceptos:** LEFT JOIN, NULL

**Pregunta:** Identifica los productos que nunca aparecen en `detalle_pedidos`. Muestra `product_id`, `product_name` y `category_id`.

````python
%sql
SELECT p.product_id,
       p.product_name,
       p.category_id
FROM productos p
LEFT JOIN detalle_pedidos d
       ON p.product_id = d.product_id
WHERE d.product_id IS NULL;
````

### Ejercicio 19. Ventas mensuales por canal

**Nivel:** Intermedio &nbsp;|&nbsp; **Conceptos:** CTE, DATE_TRUNC, agregación

**Pregunta:** Calcula las ventas netas mensuales por canal para los pedidos **no cancelados**. Muestra el mes, el canal, la cantidad de pedidos y las ventas netas.

````python
%sql
WITH ventas_pedido AS (
    SELECT o.order_id,
           DATE_TRUNC('MONTH', o.order_date) AS mes,
           o.channel,
           SUM(d.quantity * d.unit_price * (1 - d.discount_pct / 100)) AS net_order_amount
    FROM pedidos o
    INNER JOIN detalle_pedidos d ON o.order_id = d.order_id
    WHERE o.status <> 'cancelado'
    GROUP BY o.order_id, DATE_TRUNC('MONTH', o.order_date), o.channel
)
SELECT CAST(mes AS DATE)              AS mes,
       channel                        AS canal,
       COUNT(*)                       AS cantidad_pedidos,
       ROUND(SUM(net_order_amount), 2) AS ventas_netas
FROM ventas_pedido
GROUP BY mes, channel
ORDER BY mes, canal;
````

### Ejercicio 20. Producto líder por categoría

**Nivel:** Intermedio &nbsp;|&nbsp; **Conceptos:** CTE, función de ventana, DENSE_RANK

**Pregunta:** Determina el producto o productos con mayor cantidad de unidades vendidas dentro de cada categoría, considerando solo pedidos **no cancelados**. Si existe empate, muestra todos los empatados.

````python
%sql
WITH unidades AS (
    SELECT c.category_name,
           p.product_id,
           p.product_name,
           SUM(d.quantity) AS unidades_vendidas
    FROM detalle_pedidos d
    INNER JOIN pedidos o    ON d.order_id = o.order_id
    INNER JOIN productos p  ON d.product_id = p.product_id
    INNER JOIN categorias c ON p.category_id = c.category_id
    WHERE o.status <> 'cancelado'
    GROUP BY c.category_name, p.product_id, p.product_name
),
ranking AS (
    SELECT *,
           DENSE_RANK() OVER (PARTITION BY category_name
                              ORDER BY unidades_vendidas DESC) AS rk
    FROM unidades
)
SELECT category_name, product_id, product_name, unidades_vendidas
FROM ranking
WHERE rk = 1
ORDER BY category_name, product_name;
````
